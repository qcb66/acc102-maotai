```python
#collect data from other resources and print it in python

data = {
    'fyear': [2019,2020,2021,2022,2023,2024], 
    'revt': [885,979,1094,1241,1409,1678],    
    'ni': [439,495,557,627,701,823],           
    'at': [2345,2580,2890,3210,3560,3980],     
    'lt': [450,480,510,540,570,600],           
    'che': [890,950,1020,1100,1180,1260],      
    'cogs': [120,135,150,165,180,195],         
    'invt': [320,350,380,410,440,470]          
}

df = pd.DataFrame(data)
print("Raw Kweichow Moutai Financial Data (2019-2024):")
print(df[['fyear', 'revt', 'ni', 'at', 'lt', 'che', 'cogs', 'invt']])
```

```python
## 2. Data Cleaning & Financial Ratio Calculation
# Data cleaning: Keep only the columns we need for analysis
df_clean = df[['fyear', 'revt', 'ni', 'at', 'lt', 'che', 'cogs', 'invt']].copy()
```

```python
# Calculate key financial ratios for Kweichow Moutai (2019-2024)
# 1. Net Profit Margin (%) = (Net Income / Total Revenue) * 100

df_clean.loc[:, 'net_profit_margin'] = (df_clean.loc[:, 'ni'] / df_clean.loc[:, 'revt']) * 100
```

```python
# 2.ROE = (Net Income / Shareholders'Equity) * 100
# Shareholders' Equity = Total Assets - Total Liabilities
df_clean.loc[:, 'equity'] = df_clean.loc[:, 'at'] - df_clean.loc[:, 'lt']
df_clean.loc[:, 'roe'] = (df_clean.loc[:, 'ni'] / df_clean.loc[:, 'equity']) * 100
print(df_clean[['ni', 'revt', 'net_profit_margin', 'equity', 'roe']])
```

```python
# 3. Debt-to-Asset Ratio (%) = (Total Liabilities / Total Assets) * 100
# 4. Inventory Turnover = Cost of Goods Sold / Average Inventory
# We use year-end inventory for simplicity (standard for annual analysis)
df_clean.loc[:, 'debt_to_asset'] = (df_clean.loc[:, 'lt'] / df_clean.loc[:, 'at']) * 100
df_clean.loc[:, 'inventory_turnover'] = df_clean.loc[:, 'cogs'] / df_clean.loc[:, 'invt']
df_clean = df_clean.set_index('fyear')
```

```python
# Display the calculated financial ratios
print("\nCalculated Key Financial Ratios for Kweichow Moutai (2019-2024):")
print(df_clean[['net_profit_margin', 'roe', 'debt_to_asset', 'inventory_turnover']].round(2))
```

```python
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS']  
plt.rcParams['axes.unicode_minus'] = False
```

```python
## 3. Data Visualization & Trend Analysis# Create a 2x2 subplot to visualize all key financial trends
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Kweichow Moutai Key Financial Performance Trends (2019–2024)', fontsize=16, y=1.02)
```

```python
# Net Profit Margin,ROE,Debt-to-Asset Ratio and Inventory Turnover Trend
axes[0,0].plot(df_clean.index, df_clean['net_profit_margin'], marker='o', linewidth=2, color='#1f77b4')
axes[0,0].set_title('Net Profit Margin (%)', fontsize=12)
axes[0,0].set_xlabel('Fiscal Year')
axes[0,0].set_ylabel('Margin (%)')
axes[0,0].grid(True)
axes[0,1].plot(df_clean.index, df_clean['roe'], marker='s', linewidth=2, color='#ff7f0e')
axes[0,1].set_title('Return on Equity (ROE, %)', fontsize=12)
axes[0,1].set_xlabel('Fiscal Year')
axes[0,1].set_ylabel('ROE (%)')
axes[0,1].grid(True)
axes[1,0].plot(df_clean.index, df_clean['debt_to_asset'], marker='^', linewidth=2, color='#2ca02c')
axes[1,0].set_title('Debt-to-Asset Ratio (%)', fontsize=12)
axes[1,0].set_xlabel('Fiscal Year')
axes[1,0].set_ylabel('Ratio (%)')
axes[1,0].grid(True)
axes[1,1].plot(df_clean.index, df_clean['inventory_turnover'], marker='D', linewidth=2, color='#d62728')
axes[1,1].set_title('Inventory Turnover', fontsize=12)
axes[1,1].set_xlabel('Fiscal Year')
axes[1,1].set_ylabel('Turnover Ratio')
axes[1,1].grid(True)
```
